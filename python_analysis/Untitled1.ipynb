```python
import pandas as pd

from idm_plot_utils import *
```

```python
setup_root_style()

INPUT = "root://cmseos.fnal.gov//store/group/lpcmetx/iDMe/Samples/Ntuples/signal_iDMmu_muonBranches_inclusive_v1/2022EE/iDMmu"
TREE_NAME = "ntuples/outT"
BRANCH = "genSigDimuon_dr"
OUTDIR = "genSigDimuon_dr_notebook_plots"

NBINS = 15
XMIN = 0.0
XMAX = 1.5
NORMALIZE = True
```

```python
files, sample_to_files, samples = load_samples(INPUT)
```

```python
hists, summary = make_histograms_by_sample(
    sample_to_files=sample_to_files,
    samples=samples,
    branch=BRANCH,
    tree_name=TREE_NAME,
    nbins=NBINS,
    xmin=XMIN,
    xmax=XMAX,
    cut="GenSigDimuon_isValid",
    normalize=NORMALIZE,
)

print_summary(summary)
```

```python
all_items = [(s, hists[s]) for s in samples]

c_all = draw_overlay(
    all_items,
    "genSigDimuon #DeltaR, all samples",
    "genSigDimuon_dr_allSamples_overlay",
    outdir=OUTDIR,
    label_func=pretty_sample_label,
    legend_title="Signal samples",
    normalize=NORMALIZE,
    logy=False,
    legend_text_size=0.018,
)

c_all.Draw()
```

```python
canvases_by_ctau = make_fixed_ctau_canvases(
    samples=samples,
    hists=hists,
    outdir=OUTDIR,
    normalize=NORMALIZE,
    logy=False,
)

canvases_by_ctau["100"].Draw()
```
