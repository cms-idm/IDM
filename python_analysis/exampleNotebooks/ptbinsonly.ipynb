```python
def eff_determine_unc(n, d):
    eff = np.divide(n, d)
    unc = hist.intervals.ratio_uncertainty(n, d, 'efficiency')
    return eff, unc


efficiency_R_lpt, unc_eff_R_lpt = eff_determine_unc(count_num_lpt, count_den)
efficiency_R_lpt_noxclean, unc_eff_R_lpt_noxclean = eff_determine_unc(count_num_lpt_removexclean, count_den)

efficiency_R_GED, unc_eff_R_GED = eff_determine_unc(count_num_GED, count_den)

efficiency_R_TOTAL, unc_eff_R_TOTAL = eff_determine_unc(counts_TOTAL, count_den)
```
