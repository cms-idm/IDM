```python
#1. Define your model

#zfit download: maybe pip install

def double_cb(x,u,sig, nl, a_l, nr, a_r,N):
    t=(x-u)/(sig)
    al = abs(a_l)
    ar = abs(a_r)

    left_tail = t < -al
    core = (t>-al) & (t<ar)
    right_tail = t > ar

    y=np.empty_like(t)

    y[core] = np.exp(-0.5*t[core]*t[core])

    A_L = (nl / al)**nl * np.exp(-0.5 * al**2)
    B_L = (nl / al) - al
    y[left_tail] = A_L / (B_L - t[left_tail])**nl

    A_R = (nr / ar)**nr * np.exp(-0.5 * ar**2)
    B_R = (nr / ar) - ar
    y[right_tail] = A_R / (B_R + t[right_tail])**nr

    return N*y
```

```python
#2. Put your data
#Limit your ange for the fit

xdata = bins_x_lpt
ydata = sum_lpt

#3. Curve fit 
from scipy.optimize import curve_fit

initial_guess = [0, 0.02177, 1, 2, 10, 30,270] #[mean, sigma, nl, al, nr, ar, N]

popt, pcov = curve_fit(double_cb, xdata, ydata, p0=initial_guess)


plt.errorbar(bins_x_lpt, sum_lpt, yerr=YERROR_lpt, fmt='o', capsize=2, ecolor='black')
plt.xlabel(r'$(p_{T}^{\mathrm{lpt}}-p_{T}^{\mathrm{Gen}})/p_{T}^{\mathrm{Gen}}$')
plt.ylabel("Entries")

plt.plot(bins_x_lpt, double_cb(bins_x_lpt, *popt), label='Double Crystal Ball Fit', color='red', linewidth=2)

yp = double_cb(bins_x_lpt, *popt)

eq = (ydata-yp)/(YERROR_lpt)
# print (eq)
chi2 = np.sum(eq**2)

ndof = len(ydata) - len(popt)

chi2ndof = chi2/ndof
# print (chi2)
# print (ndof)
# print (chi2ndof)
print (popt)
```
