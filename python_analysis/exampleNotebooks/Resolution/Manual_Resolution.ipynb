## 3. $L_{xy}$ resolution 

This study needs a different approach from the electron $p_T$ and $d_{xy}$ resolution studies. The quantities currently stored as gen and reco electron `vxy` are defined differently:

- **Gen electron:** `nt.genEleVxy_ = genParticle.vertex().rho();` gives the transverse position of the generated electron’s production vertex.
- **Reco low-$p_T$ electron:** `nt.recoAllLowPtElectronVxy_.push_back(ele.trackPositionAtVtx().rho());` gives the transverse position of the electron’s track at its vertex. It is an electron quantity, not the position of a fitted dielectron vertex.

A direct comparison of these two stored electron quantities would therefore not measure the resolution of the reconstructed **dielectron vertex**.

Instead, we match the reconstructed dielectron vertex to the generated decay vertex and compare their transverse displacements. This tells us how accurately we reconstruct the vertex position. The matched reconstructed vertex can be of type **LL** or **RR**.

We define the relative $L_{xy}$ residual as

$$
\frac{L_{xy}^{\mathrm{reco\ vertex}} - L_{xy}^{\mathrm{gen\ vertex}}}
     {L_{xy}^{\mathrm{gen\ vertex}}}.
$$

The width of this residual distribution provides a measure of the $L_{xy}$ resolution.

The low-$p_T$ and GED algorithms show similar resolution in the studies above

Since the low-pT and the GED algorithms show similar resolutions, and the low-pT demonstrates better and more consistent efficiency (as shown in my efficiency studies), I think it is reasonable to move forward using only the low-pT algorithm.
