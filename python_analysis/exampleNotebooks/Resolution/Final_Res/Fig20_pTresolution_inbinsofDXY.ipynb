```python
import numpy as np
import matplotlib.pyplot as plt

# pT binning
lpt_genpt_edges = np.array([0,0.2,0.4,0.6,0.8,1,1.2,1.4,2])
ged_genpt_edges = np.array([0,0.2,0.4,0.6,0.8,1,1.2,1.4,2])

lpt_genpt = (lpt_genpt_edges[:-1] + lpt_genpt_edges[1:]) / 2
ged_genpt = (ged_genpt_edges[:-1] + ged_genpt_edges[1:]) / 2

# Final Low-pT fit results
lpt_sigma= np.array([0.021113154961413545, 0.019975160219183708, 0.018259910222920544, 0.019098961220072017, 0.01946821017384812, 0.019056905047694934, 0.018220030724722735, 0.018769296692508694])
lpt_sigma_unc= np.array([0.00024354741364826496, 0.0003448539909220785, 0.00043415084624029963, 0.00048182629361454693, 0.0006684876279429914, 0.000674383179229603, 0.0007871034201864417, 0.0018863555567394101])

ged_sigma= np.array([0.020830913636631154, 0.019384653186779587, 0.02035025964579439, 0.018813415650755155, 0.020743934060980893, 0.020493415420702103, 0.018965156323850648, 0.018202926972437992])
ged_sigma_unc= np.array([0.00026996875162781255, 0.0004046770450463634, 0.0004133999678855084, 0.0006221893093095231, 0.0006480485844486072, 0.0008414609091423402, 0.0009225432415505677, 0.0025422262506062483])
```

```python
fig, ax = plt.subplots(figsize=(8, 8))

ax.stairs(
    lpt_sigma,
    lpt_genpt_edges,
    baseline=None,
    color="C0",
)

ax.errorbar(
    lpt_genpt,
    lpt_sigma,
    yerr=lpt_sigma_unc,
    fmt="o",
    color="C0",
    capsize=2,
    label="Low-$p_T$"
)

ax.stairs(
    ged_sigma,
    ged_genpt_edges,
    baseline=None,
    color="C1",
)

ax.errorbar(
    ged_genpt,
    ged_sigma,
    yerr=ged_sigma_unc,
    fmt="s",
    color="C1",
    capsize=2,
    label="GED"
)

ax.set_xlabel(r"$d_{xy}$ [cm]")
ax.legend()
ax.grid(True, linestyle="--", alpha=0.2)

plt.ylabel(
    r"DSCB $\sigma$ of $(p_{T}^{\mathrm{reco}}-p_{T}^{\mathrm{Gen}})/p_{T}^{\mathrm{Gen}}$"
)

plt.ylim(0, 0.08)
plt.show()
```
