```python
import numpy as np
import matplotlib.pyplot as plt

# pT binning
lpt_genpt_edges = np.array( [0.0, 2.0, 5.0, 10.0] )
ged_genpt_edges = np.array( [0.0, 2.0, 5.0, 10.0] )

lpt_genpt = (lpt_genpt_edges[:-1] + lpt_genpt_edges[1:]) / 2
ged_genpt = (ged_genpt_edges[:-1] + ged_genpt_edges[1:]) / 2

# Final Low-pT fit results
lpt_sigma= np.array( [0.005938598464236792, 0.008594394232356823, 0.02776306072803079] )
lpt_sigma_unc= np.array( [0.00011486758470279072, 0.00011734939817231783, 0.0006152662812668739] )

ged_sigma= np.array([0.005878948620613452, 0.008556710120996838, 0.027697174828691946])
ged_sigma_unc= np.array([0.000118523378390651, 0.00011791186917821954, 0.000609880308218029])
```

```python
fig, ax = plt.subplots(figsize=(8, 8))

ax.stairs(
    lpt_sigma,
    lpt_genpt_edges,
    baseline=None,
    color="C0",
)

ax.errorbar(
    lpt_genpt,
    lpt_sigma,
    yerr=lpt_sigma_unc,
    fmt="o",
    color="C0",
    capsize=2,
    label="Low-$p_T$"
)

ax.stairs(
    ged_sigma,
    ged_genpt_edges,
    baseline=None,
    color="C1",
)

ax.errorbar(
    ged_genpt,
    ged_sigma,
    yerr=ged_sigma_unc,
    fmt="s",
    color="C1",
    capsize=2,
    label="GED"
)

ax.set_xlabel(r"$L_{xy}$ [cm]")
ax.legend()
ax.grid(True, linestyle="--", alpha=0.2)

plt.ylabel(
    r"DSCB $\sigma$ of $(d_{xy}^{\mathrm{reco}}-d_{xy}^{\mathrm{Gen}})/d_{xy}^{\mathrm{Gen}}$"
)

plt.ylim(0, 0.08)
plt.show()
```

```python
difference = (lpt_sigma - ged_sigma) / ged_sigma * 100

plt.figure(figsize=(8,6))

plt.plot(
    lpt_genpt,
    difference,
    'o-'
)

plt.axhline(0, linestyle='--')

plt.xlabel(r"$p_T$ [GeV]")
plt.ylabel(r"$(\sigma_{\mathrm{LPT}}-\sigma_{\mathrm{GED}})/\sigma_{\mathrm{GED}}$ [%]")

plt.show()
```
