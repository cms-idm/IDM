```python
import numpy as np
import matplotlib.pyplot as plt

# pT binning
lpt_genpt_edges = np.array([0, 1, 3, 6, 15])
ged_genpt_edges = np.array([0, 1, 3, 6, 15])

lpt_genpt = (lpt_genpt_edges[:-1] + lpt_genpt_edges[1:]) / 2
ged_genpt = (ged_genpt_edges[:-1] + ged_genpt_edges[1:]) / 2

# Final Low-pT fit results
lpt_sigma = np.array([0.020119, 0.020289, 0.021004, 0.022811])
lpt_sigma_unc = np.array([0.000151, 0.000209, 0.000446, 0.001071])

# Final GED fit results
ged_sigma = np.array([0.020742, 0.019985, 0.019993,0.020878 ])
ged_sigma_unc = np.array([0.000173, 0.000204, 0.000407, 0.000995	])
```

```python
fig, ax = plt.subplots(figsize=(8, 8))

ax.stairs(
    lpt_sigma,
    lpt_genpt_edges,
    baseline=None,
    color="C0",
)

ax.errorbar(
    lpt_genpt,
    lpt_sigma,
    yerr=lpt_sigma_unc,
    fmt="o",
    color="C0",
    capsize=2,
    label="Low-$p_T$"
)

ax.stairs(
    ged_sigma,
    ged_genpt_edges,
    baseline=None,
    color="C1",
)

ax.errorbar(
    ged_genpt,
    ged_sigma,
    yerr=ged_sigma_unc,
    fmt="s",
    color="C1",
    capsize=2,
    label="GED"
)

ax.set_xlabel(r"$L_{xy}$ [cm]")
ax.legend()
ax.grid(True, linestyle="--", alpha=0.2)

plt.ylabel(
    r"DSCB $\sigma$ of $(p_{T}^{\mathrm{reco}}-p_{T}^{\mathrm{Gen}})/p_{T}^{\mathrm{Gen}}$"
)
# plt.xlim(-0.003,15)
plt.ylim(0, 0.08)
plt.show()
```
