# Resolution Studies


Aim: 
To evaluate and compare the resolution performance of the GED and low-pT electron reconstruction algorithms. 
Based on this comparison, we will determine whether retaining GED electrons provides a significant benefit to the analysis or whether the low-pT algorithm alone is sufficient.


## How do we plan to do this?

We will study:

- $p_T$ resolution in bins of gen $p_T$, gen $L_{xy}$, and gen $d_{xy}$
- $d_{xy}$ resolution in bins of gen $p_T$ and gen $L_{xy}$
- $L_{xy}$ resolution
