# Electron Reconstruction Efficiency


Electron reconstruction efficiency (ERE) is calculated as the fraction of generator-level signal electrons matched to a reconstructed GED or low-$p_T$ electron within $\Delta R < 0.1$.

$$
\text{Electron Reconstruction Efficiency}
=
\frac{
N\left(\text{generator-level signal electrons matched to a reconstructed GED or low-}p_T\text{ electron with }\Delta R < 0.1\right)
}{
N\left(\text{total generator-level signal electrons}\right)
}
$$

We study efficiency as a function of pT and Lxy- two very imp features of IDM.
